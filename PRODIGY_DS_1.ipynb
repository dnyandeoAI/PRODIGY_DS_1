```python
# Select the country
country = 'India'

# Select every 10th year from 1960 to 2020 (or 2024 if you want)
year_cols = [str(year) for year in range(1960, 2025, 10)]

# Get population values for those years
data_country = df[df['Country Name'] == country]
population = data_country[year_cols].iloc[0]

# Convert years to integers for x-axis
years = list(map(int, year_cols))

# Make a bigger figure
plt.figure(figsize=(12,6))  # bigger plot

# Bar plot
plt.bar(years, population, color='skyblue', width=5)
plt.xlabel('Year')
plt.ylabel('Population')
plt.title(f'Population of {country} (Every 10 Years)')
plt.xticks(years, rotation=45)
plt.show()
```

```python
# Select the country
country = 'China'

# Select every 10th year from 1960 to 2024
year_cols = [str(year) for year in range(1960, 2025, 10)]

# Filter China data
china_data = df[df['Country Name'] == country]

# Get values for the selected years
values = china_data[year_cols].iloc[0]

# Convert years to integers for plotting
years = list(map(int, year_cols))
```

```python
plt.figure(figsize=(12,6))  # bigger figure

plt.bar(years, values, color='orange', width=5)
plt.xlabel('Year')
plt.ylabel('Population')
plt.title(f'Population of {country} (Every 10 Years)')
plt.xticks(years, rotation=45)
plt.show()
```

```python
growth_pct = values.pct_change() * 100
print(f"Decade-wise growth (%) for {country}:\n", growth_pct)
```

```python
max_pop = values.max()
min_pop = values.min()
print(f"Max Population: {max_pop}, Min Population: {min_pop}")
```

```python
other_country = 'India'
india_values = df[df['Country Name'] == other_country][year_cols].iloc[0]

plt.figure(figsize=(12,6))
plt.bar([y - 1 for y in years], india_values, width=2, label='India', color='skyblue')
plt.bar([y + 1 for y in years], values, width=2, label='China', color='orange')
plt.xlabel('Year')
plt.ylabel('Population')
plt.title('Population Comparison: China vs India (Every 10 Years)')
plt.xticks(years, rotation=45)
plt.legend()
plt.show()
```

```python
countries = ['China', 'India', 'United States', 'Brazil']

# Every 10th year from 1960 to 2024
years = [str(year) for year in range(1960, 2025, 10)]

# Filter dataset for selected countries and years
data_countries = df[df['Country Name'].isin(countries)].set_index('Country Name')[years]

# Convert years to integers for plotting
years_int = list(map(int, years))

# Plotting
plt.figure(figsize=(14,7))
bar_width = 2

for i, country in enumerate(countries):
    # Shift bars slightly to avoid overlap
    plt.bar([y + i*bar_width for y in years_int], data_countries.loc[country], 
            width=bar_width, label=country)

plt.xlabel('Year (Decade)')
plt.ylabel('Population')
plt.title('Population Comparison: China, India, USA, Brazil (Every 10 Years)')
plt.xticks(years_int, rotation=45)
plt.legend()
plt.show()
```

```python
# Rename columns and merge with your main dataset
metadata_renamed = metadata.rename(columns={'AFE': 'Country Code', 'Unnamed: 1': 'Region'})
df_merged = pd.merge(df, metadata_renamed[['Country Code', 'Region']], on='Country Code', how='left')

# Keep only rows that have a Region (i.e., real countries)
df_countries = df_merged[df_merged['Region'].notna()]
```

```python
top10_2024 = df_countries[['Country Name', '2024']].copy()
top10_2024['2024'] = pd.to_numeric(top10_2024['2024'], errors='coerce')
top10_2024 = top10_2024.sort_values(by='2024', ascending=False).head(10)

print(top10_2024)
```

```python
# Bar chart
plt.figure(figsize=(12,6))
plt.bar(top10_2024['Country Name'], top10_2024['2024'], color='skyblue')
plt.xlabel('Country')
plt.ylabel('Population (2024)')
plt.title('Top 10 Most Populated Countries in 2024')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()  # Adjust layout so labels fit
plt.show()
```

```python
top10_2014 = df_countries[['Country Name', '2014']].copy()
top10_2014['2014'] = pd.to_numeric(top10_2014['2014'], errors='coerce')
top10_2014 = top10_2014.sort_values(by='2014', ascending=False).head(10)
print(top10_2014)
```

```python
# Bar chart
plt.figure(figsize=(12,6))
plt.bar(top10_2014['Country Name'], top10_2014['2014'], color='orange')
plt.xlabel('Country')
plt.ylabel('Population (2014)')
plt.title('Top 10 Most Populated Countries in 2014')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Merge 2014 and 2024
comparison = pd.merge(top10_2014, top10_2024, on='Country Name')

# Side-by-side bar plot
x = np.arange(len(comparison['Country Name']))
width = 0.35

plt.figure(figsize=(14,7))
plt.bar(x - width/2, comparison['2014'], width, label='2014', color='lightcoral')
plt.bar(x + width/2, comparison['2024'], width, label='2024', color='skyblue')

plt.xlabel('Country')
plt.ylabel('Population')
plt.title('Population Comparison: Top 10 Countries (2014 vs 2024)')
plt.xticks(x, comparison['Country Name'], rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
df_2024 = df_countries[['Country Name', '2024']].copy()
df_2024['2024'] = pd.to_numeric(df_2024['2024'], errors='coerce')
bottom10_2024 = df_2024.sort_values(by='2024', ascending=True).head(10)
bottom10_2024
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
    data=bottom10_2024, 
    y='Country Name', 
    x='2024', 
    palette='viridis',
    dodge=False
)
plt.title('Top 10 Least Populated Countries in 2024', fontsize=16, weight='bold')
plt.xlabel('Population', fontsize=12)
plt.ylabel('Country', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# Create a copy with 2014 column
df_2014 = df_countries[['Country Name', '2014']].copy()
df_2014['2014'] = pd.to_numeric(df_2014['2014'], errors='coerce')

# Filter only the top 10 least populated countries from 2024
bottom10_2014 = df_2014[df_2014['Country Name'].isin(bottom10_2024['Country Name'])]

# Sort by 2014 population for nicer plotting
bottom10_2014 = bottom10_2014.sort_values(by='2014', ascending=True)
bottom10_2014
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
    data=bottom10_2014, 
    y='Country Name', 
    x='2014', 
    palette='coolwarm',
    dodge=False
)
plt.title('Top 10 Least Populated Countries in 2014', fontsize=16, weight='bold')
plt.xlabel('Population', fontsize=12)
plt.ylabel('Country', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
countries = bottom10_2024['Country Name']
pop_2014 = bottom10_2014['2014']
pop_2024 = bottom10_2024['2024']

plt.figure(figsize=(12,7))
plt.plot(countries, pop_2014, marker='o', linestyle='--', color='skyblue', label='2014')
plt.plot(countries, pop_2024, marker='o', linestyle='--', color='limegreen', label='2024')

plt.xticks(rotation=45)
plt.ylabel('Population')
plt.title('Population Comparison: 2014 vs 2024 (Least Populated Countries)', fontsize=16, weight='bold')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Top 10 most populated countries in 2024
top10_2024 = df_countries.sort_values('2024', ascending=False).head(10)

plt.figure(figsize=(10,10))
plt.pie(
    top10_2024['2024'], 
    labels=top10_2024['Country Name'], 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=plt.cm.tab20.colors
)
plt.title('Top 10 Most Populated Countries in 2024', fontsize=16, weight='bold')
plt.show()
```

```python
# Top 10 most populated countries
top10 = df_countries.sort_values('2024', ascending=False).head(10)

# Calculate rest of world population
rest_population = df_countries['2024'].sum() - top10['2024'].sum()

# Combine top 10 + rest of world
labels = list(top10['Country Name']) + ['Rest of the World']
sizes = list(top10['2024']) + [rest_population]

plt.figure(figsize=(10,10))
plt.pie(
    sizes, 
    labels=labels, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=plt.cm.tab20.colors
)
plt.title('Population Distribution: Top 10 Countries vs Rest of the World (2024)', fontsize=16, weight='bold')
plt.show()
```
